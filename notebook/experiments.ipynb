```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
datalist=[]
p=Path("data").resolve()
for file in p.glob("*.pdf"):
    datalist.extend(PyPDFLoader(file).load())
documents=datalist    
loader=PyPDFLoader(file_path)

```

```python
documents=loader.load()
```

```python
documents
```

```python
len(documents)
```

```python
text_splitter=RecursiveCharacterTextSplitter(
 chunk_size=500,
 chunk_overlap=150,
 length_function=len   
)
```

```python
docs=text_splitter.split_documents(documents)
```

```python
len(docs)
```

```python
docs[0].metadata
```

```python
docs[0].page_content
```

THIS IS RETRIEVAL PROCESS. 


```python
from langchain_community.vectorstores import FAISS
```

```python
vectorstore=FAISS.from_documents(docs,embedding_model)
```

```python
relevant_doc=vectorstore.similarity_search("nanomaterials")
```

```python
relevant_doc[0].page_content
```

```python
retriever=vectorstore.as_retriever(search_kwags={"k":10})
```

```python
prompt_sample="""Answer the user question based off of the context. If there is not much 
                tell the user that you do not know the answer and that you need more information.
                Context: {Context}

                Question: {Question}

                Answer:
                """
```

```python
from langchain_classic import PromptTemplate
prompt= PromptTemplate(
    template=prompt_sample,
    input_variables=["Context","Question"]
)
```

```python
prompt
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser=StrOutputParser()
```

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
rag_chain= (
    {"Context": retriever|format_docs , "Question": RunnablePassthrough()}
    |    prompt
    | llm
    | parser #lcel language langchain expression language
)
```

```python
rag_chain.invoke("waht are these documents about nanomaterials")
```
