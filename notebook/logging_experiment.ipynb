```python
import logging
import os 
from datetime import datetime
```

```python
log_dir=os.path.join(os.getcwd(),"logs")
```

```python
os.makedirs(log_dir, exist_ok=True)
```

```python
LOG_FILE=f"{datetime.now().strftime('%m_%d_%Y_%H_%M_%S')}.log" #this is just creating a string with that name. f string is appending to .log
```

```python
#this is a log file ( name of the log file is current date and time)
LOG_FILE
```

```python
LOG_FILE_path=os.path.join(log_dir, LOG_FILE) #this also does not create a file. this constructs the corrrect path for the os. This line simply joins two strings to form a file path.
```

the basic config line
This line configures the logging system to write logs to the file at LOG_FILE_PATH.

Whether the file is created immediately depends on the logging module's behavior. In practice:

logging.basicConfig(...) sets up the file handler.
The log file is typically created when the first log record is written.

```python
logging.basicConfig(
    filename=LOG_FILE_path,
    format="[ %(asctime)s ] %(levelname)s %(name)s (line:%(lineno)d) - %(message)s ", # "%" is a format specifier. asctime gives the current time. lineeno gives the first line and then we print the message. The level is set to info the previous levels cannot be called the next levels can be called 
    level=logging.INFO,
)
```

.getlogger line
does not create a file. It creates (or retrieves) a Logger object with the name
What getLogger() does

The logging module can have multiple loggers, each with its own name.

```python
logger=logging.getLogger("DocumentPortal")
```

Here's what happens:

basicConfig() configures the logging system to write to app.log.
getLogger("DocumentPortal") returns a logger named "DocumentPortal".
logger.info(...) sends a log record through the configured logging system.
The logging system writes that record to log_file_path

```python
logger.info("This is an info message")
```
